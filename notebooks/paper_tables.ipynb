```python
import numpy as np
import PyWVLTConCoeff as cc
import csv
import pywt
```

## Generate db3 filter table

```python
filt = np.asarray(pywt.Wavelet('db3').filter_bank[2])

with open('../data/db3_filt.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(filt)
```

## Generate supplementary material connection coefficient table

```python
cc = cc.coeffs.four_term_coeff_solver('db3', np.array([0, 1, 2]))

idx = np.arange(-4, 5)
brk = 9 * 6

tmp_k1, tmp_k2, tmp_k3 = np.meshgrid(idx, idx, idx, indexing='ij')
k1 = tmp_k1.flatten()
k2 = tmp_k2.flatten()
k3 = tmp_k3.flatten()

# Generate CSV files
j = 1
for i in range(0, 12, 3):
    k11 = k1[i*brk:(i+1)*brk]
    k21 = k2[i*brk:(i+1)*brk]
    k31 = k3[i*brk:(i+1)*brk]
    k12 = k1[(i+1)*brk:(i+2)*brk]
    k22 = k2[(i+1)*brk:(i+2)*brk]
    k32 = k3[(i+1)*brk:(i+2)*brk]
    k13 = k1[(i+2)*brk:(i+3)*brk]
    k23 = k2[(i+2)*brk:(i+3)*brk]
    k33 = k3[(i+2)*brk:(i+3)*brk]
    cc1 = cc[i*brk:(i+1)*brk]
    cc2 = cc[(i+1)*brk:(i+2)*brk]
    cc3 = cc[(i+2)*brk:(i+3)*brk]
    data1 = {'ka': k11, 'kb': k21, 'kc': k31, 'cca': cc1,
             'kd': k12, 'ke': k22, 'kf': k32, 'ccb': cc2,
             'kg': k13, 'kh': k23, 'ki': k33, 'ccc': cc3}

    with open(f'../data/db3_012_pt{j}.csv', 'w', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(data1.keys())
        writer.writerows(zip(*data1.values()))
    f.close()

    j += 1

# Handle final part
i += 3
tmp = i * brk
k11 = k1[tmp:tmp + 3*9]
k21 = k2[tmp:tmp + 3*9]
k31 = k3[tmp:tmp + 3*9]
cc1 = cc[tmp:tmp + 3*9]
tmp += 3*9
k12 = k1[tmp:tmp + 3*9]
k22 = k2[tmp:tmp + 3*9]
k32 = k3[tmp:tmp + 3*9]
cc2 = cc[tmp:tmp + 3*9]
tmp += 3*9
k13 = k1[tmp:]
k23 = k2[tmp:]
k33 = k3[tmp:]
cc3 = cc[tmp:]
data1 = {'ka': k11, 'kb': k21, 'kc': k31, 'cca': cc1,
         'kd': k12, 'ke': k22, 'kf': k32, 'ccb': cc2,
         'kg': k13, 'kh': k23, 'ki': k33, 'ccc': cc3}

with open(f'../data/db3_012_pt{j}.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(data1.keys())
    writer.writerows(zip(*data1.values()))
f.close()
```
