```python
import matplotlib.pyplot as plt
import numpy as np
import wvltcc
```

```python
deriv = np.array([0, 1, 2])
wvtl = 'db3'

four_term012 = wvltcc.coeffs.four_term_coeff_solver(wvtl, deriv)
three_term01 = wvltcc.coeffs.three_term_coeff_solver(wvtl, np.array([0, 1]))
three_term12 = wvltcc.coeffs.three_term_coeff_solver(wvtl, np.array([1, 2]))
three_term02 = wvltcc.coeffs.three_term_coeff_solver(wvtl, np.array([0, 2]))
two_term0 = wvltcc.coeffs.two_term_coeff_solver(wvtl, 0)
two_term1 = wvltcc.coeffs.two_term_coeff_solver(wvtl, 1)
two_term2 = wvltcc.coeffs.two_term_coeff_solver(wvtl, 2)

coeffs = {
    'four_term012': four_term012,
    'three_term01': three_term01,
    'three_term12': three_term12,
    'three_term02': three_term02,
    'two_term0': two_term0,
    'two_term1': two_term1,
    'two_term2': two_term2
}
```

```python
keys = list(coeffs.keys())

for i, key in enumerate(keys):
    fig, ax = plt.subplots(figsize=(6, 4))
    c = coeffs[key]
    ax.plot(c, color="#2b8057")
    ax.set_xticks([])
    ax.set_yticks([])
    fig.tight_layout()
    fig.savefig(f'../figures/{key}.png', dpi=100)
```
