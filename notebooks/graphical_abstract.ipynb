```python
import matplotlib.pyplot as plt
import numpy as np
import PyWVLTConCoeff as cc
```

```python
deriv = np.array([0, 1, 1])
wvtl = 'db3'

four_term011 = cc.coeffs.four_term_coeff_solver(wvtl, deriv)
three_term01 = cc.coeffs.three_term_coeff_solver(wvtl, np.array([0, 1]))
three_term11 = cc.coeffs.three_term_coeff_solver(wvtl, np.array([1, 1]))
two_term0 = cc.coeffs.two_term_coeff_solver(wvtl, 0)
two_term1 = cc.coeffs.two_term_coeff_solver(wvtl, 1)

coeffs = {
    'four_term011': four_term011,
    'three_term01': three_term01,
    'three_term11': three_term11,
    'two_term0': two_term0,
    'two_term1': two_term1
}
```

```python
keys = list(coeffs.keys())

for i, key in enumerate(keys):
    fig, ax = plt.subplots(figsize=(6, 4))
    c = coeffs[key]
    ax.plot(c, color="#2b8057")
    ax.set_xticks([])
    ax.set_yticks([])
    fig.tight_layout()
    fig.savefig(f'../figures/{key}.png', dpi=100)
```
