```python
import numpy as np
import PyWVLTConCoeff as cc
```

## Run utils for recreating the error plots in the paper

```python
wvlt = ['db3']

for w in wvlt:
    # plot two-term residuals
    cc.utils.plot_two_term_residuals(w, deriv_orders=np.arange(4), path='../figures', font_size=16)

    # # plot three-term residuals
    x, y = np.meshgrid(np.arange(4), np.arange(4), indexing='ij')
    deriv_orders = np.vstack([x.ravel(), y.ravel()])
    cc.utils.plot_three_term_residuals(w, deriv_orders=deriv_orders, path='../figures', font_size=16)

    # plot four-term residuals
    x, y, z = np.meshgrid(np.arange(4), np.arange(4), np.arange(4), indexing='ij')
    deriv_orders = np.vstack([x.ravel(), y.ravel(), z.ravel()])
    cc.utils.plot_four_term_residuals(w, deriv_orders=deriv_orders, path='../figures', font_size=16)

    # plot four-term residuals up to a certain order
    cc.utils.plot_four_term_residuals_up_to(w, deriv_orders=deriv_orders, max_order=2, path='../figures', font_size=16)
```
